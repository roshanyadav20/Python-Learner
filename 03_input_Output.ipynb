**Basic Input & Output**

**print() function**: display ouput to screen

```python
name = "Roshan"
print("Name:",name)
```

**input() function**: take input from user and storing input in variables.

```python
name = input("Enter your name: ")
print("Name:",name)
```

```python
age = input("Enter your age: ")
print("My age is:",age)
```

*The input() function always returns a string.*

```python
print("Type of age",type(age)) #
```

**Converting input types**: int(), float(), etc

**1. First method**: Input and type conversion in **two** steps.

```python
age = input("Enter your age: ")
age = int(age) # convert the input to an integer
print("My age is:",age)
print("Type of age:",type(age))
```

**Second method**: Input and type conversion in **one** step.

```python
age = int(input("Enter your age: "))
print("My age is:",age)
print("Type of age:",type(age))
```

**Without Formatting output example**

```python
name = "Roshan"
print(name)
```

**Formatting Output using**:

*   Commas (,)
*   f-string
*   format() method

**1. Commas (,)**: Commas are not a recommended method for formatting output in Python.

```python
name = "Roshan"
print("My name is:",name)
```

**2. f-string**: f-string is the most recommended method for formatting output in Python.

```python
name = "Roshan"
age = 23
# f-string ka use
print(f"My name is {name} and my age is {age}")
```

**3. format() method**: is used for formatting output, but it is not as popular as f-strings.

```python
name = "Roshan"
age = 23
# format() method ka use.
print("My Name is {} and my age is {}".format(name,age))
```
