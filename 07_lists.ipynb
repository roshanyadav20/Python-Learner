**Lists in Python:** A List is a data structure used to store multiple items in a single variable.

**Without using Lists**

```python
# This is not an efficient(good) way to store multiple values, using multiple variables.
num1 = 10
num2 = 20
num3 = 30
num4 = 40
num5 = 50

print(num1, num2, num3, num4, num4)
```

**1. Creating Lists:**

```python
num = [1,3,4,7]
print(num)
print(type(num))
```

```python
num = [1, "Roshan", 4, 7, 10, True]
print(num)
print(type(num))
```

**Empty List:** An empty list is a valid list. A list can be created without storing any items.

```python
num = []
print(num)
print(type(num))
```

**2. Accessing List items:**

```python
num = [10,20,45,16]
print(num[0]) # acess first element
print(type(num))
```

```python
print(num[3]) # acess fourth element
```

*Lenght of the list*

```python
num = [10,20,45,16,24]
print(len(num))
```

**3. Changing List Items:** We can change a list item by using its index.


*   Lists are mutable, which means their items can be changed.
*   **Indexing** in a list starts from **0**.

```python
fruits = ["apple","banana","cherry","date","grapes"]
fruits[1] = "blueberry"
print(fruits)
```

```python
fruits[0] = "blueberry"
print(fruits)
```

**Note Points:** Strings are Immutable (Cannot be Changed), You cannot change a string character using its index.

*   name = "Sohan"
*   print(name[0]) = "M"
*   Output: TypeError



**4. Slicing in List:** It is used to get a specific part of a list.

*   list[start : end : step]
*   The start index is counted.
*   the end index is not counted.
*   step defines how many items to skip.

```python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[1:4])
print(fruits[:4])
print(fruits[1:])
print(fruits[-3:-1])
```

**5. Useful List Methods:** are predefined functions.
*   **append() - List method:** append() adds one element at the end of a list, but it does not return the added element.

```python
num = [10,30,80,12]
print(num.append(50)) # not return the added element.
```

```python
# To return the added element, use print() function with the list name.
num = [10,30,80,12]
num.append(50)
print(num)
```

```python
# remove()
num = [10,30,80,12]
num.remove(80)
print(num)
```

```python
# clear() - puri list ki value delete karna.
num = [10,30,80,12]
num.clear()
print(num)
```

```python
# pop() - any specefic(only one) value delete through index number.
num = [10,30,80,12]
num.pop(2)
print(num)
```

```python
# insert
num = [10,30,80,12]
num.insert(2, 100) # (index, new value add)
print(num)
```

```python
# sort() -
num = [10,30,80,12] # number
num.sort() # sort list in ascending Order.
print(num)
```

```python
# sort()- string mein bhi sorting kar sakte hai.
num = ["a","c", "b", "d"]
num.sort()
print(num)
```

```python
# Descending Order - sort(reverse=True)
num = ["a","c", "b", "d"]
num.sort(reverse=True)
print(num)
```
