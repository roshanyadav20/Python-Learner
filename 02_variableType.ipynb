**Python Data Types in Python**

***1. Number Data Type***

```python
# 1. Integer
age = 30
print("Age:", age)
print(type(age))
```

```python
# 2. Float
pi = 3.14
print("Pi:", pi)
print(type(pi))
```

***2. String Data Type***

```python
name = "Roshan"
print("Name:", name)
print(type(name))
```

***3. Boolean***

```python
isStudent = True
print("Is student:", isStudent)
print(type(isStudent))
```

**Dynamic typing in Python:-** Dynamic typing means Python automatically decides the data type of a variable based on the value assigned to it

```python
x = 10
print(type(x))

x = "Roshan"
print(type(x))

x = 10.5
print(type(x))
```

*Interview Question*

```python
nothing = None
print("None:",nothing)
print(type(nothing))
```

**1. Type Conversion (Implicit)**: Python automatically converts one data type into another.

```python
x = 10 # int
y = 20.5 # float
print(x+y) # Output: float || Automatically by Python.
```

```python
x = 10
y = 20.5
z = x + y
print(z)
print(type(z))
```

**2. Type Casting in Python**: We forcefully convert one data type into another data type using functions like int(), float(), and str().

```python
x = "10" #string
a = int(x) #int
print(a)
print(type(a))
```

```python
x = int("10")
print(x)
print(type(x))
```

**Remember**: String conversion depends on the value stored inside the string.

```python
# String with only number.
a = "25"
print(int(a))    # 25
```

```python
# String with only float.
a = "25.5"
print(float(a))  # 25.0
```
