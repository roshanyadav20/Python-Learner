**Conditional Statements in Python:** Conditional statements allows your program to make decisions- run different parts of code based on certain condiitons.

*   A condiiton is simply a statement that can be either True or False.

Types of Conditional Statement:-

*   if statement
*   if-else statement
*   if-elif-else statement
*   Nested if statement

**Boolean Expression:** A condition can return True or False without using an if statement.

```python
age = 20
print(age >= 18)
```

```python
age = 16
print(age >= 18)
```

**1. if statement:** is used to run a block of code only when the condition is True.

```python
age = 20
if age >= 18:
  print("You are eligible for vote")
```

```python
# No output (because the condition is False).
age = 16
if age >= 18:
  print("You are eligible for vote")
```

**2. if-else statement:** If the condition is True, if runs; if the condition is False then, else runs.

```python
# If the condition is True, (if) runs.
age = 20
if age >= 18:
  print("You can give vote.")
else:
  print("You can not give vote.")
```

```python
# if the condition is False then, (else) runs.
age = 16
if age >= 18:
  print("You can give vote.")
else:
  print("You can not give vote.")
```

```python
if (True):
  print("You can give vote.")
else:
  print("You can not give vote.")
```

```python
if (False):
  print("You can give vote.")
else:
  print("You can not give vote.")
```

**3. if-elif-else statement:** It is used when we have multiple conditions.

```python
marks = 95
if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Grade D")
```

```python
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Grade D")
```

```python
marks = 65
if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Grade D")
```

**Practice Problem:** # Write a Python program that takes a number as input and prints:

```python
num = int(input("Enter a number: "))
if (num > 0):
  print("Positive")
elif (num == 0):
  print("Zero")
else:
  print("Negative")
```

**4. Nested if statement:** It is used when an if statement is written inside another if statement.

```python
# Write a Python program to check whether a number is positive and even.
num = 10
if num > 0:
  print("Positive")
  if num % 2 == 0:
    print("Even")
```
