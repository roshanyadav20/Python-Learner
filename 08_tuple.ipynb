**1. Creating Tuples:** A tuple is a data structure used to store multiple items in a single variable.

*   In a tuple, use commas (,) between items. A single item also needs a comma.
*   Tuples are created using ().
*   Indexing starts from 0.
*   Tuples allow duplicate items.
*   Tuples are immutable (cannot be changed).
*   Tuples can store different data types.
*   Tuples do not support append().

*Not a tuple*

```python
# not a tuple
num = (10)
print(num)
print(type(num))
```

*Create tuple with one item*

```python
num = (10,)
print(num)
print(type(num))
```

*Create tuple with multiple item*

```python
fruits = ("Apple", "banana", "cherry")
print(fruits)
print(type(fruits))
```

*length of the tuple*

```python
fruits = ("Apple", "banana", "cherry")
print(len(fruits))
```

*Tuples can also contain different data types.*

```python
num = (10,13,15,7.5,"Roshan",True,False)
print(num)
print(type(num))
```

```python
fruits = ("Apple", "Banana", "Mango")

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

**Slicing in Tuple:**

*   Acess tuple item through negative indexing



*   Acess tuple items through range of indexes.

```python
fruits = ("Apple", "Banana", "Mango", "cherry", "grapes")
print(fruits[2:5])
```

**check if the item exists:**

```python
fruits = ("Apple", "Banana", "Mango", "cherry", "grapes")
if "Banana" in fruits:
  print("Yes, 'Banana' is in the fruits tuple")
```

**(Very important) Changing of tuple values**

```python
fruits = ("Apple", "Banana", "Mango", "cherry", "grapes")
print(type(fruits))
y = list(fruits) #list
print(type(y))

y[1] = "Python"
fruits = tuple(y)
print(fruits)
print(type(fruits))
```

*Add items in tuple(very inportant)*

```python
fruits = ("Apple", "Banana", "Mango", "cherry", "grapes")
print(type(fruits))
y = list(fruits)
print(type(y))

y.append("last value add")
fruits = tuple(y)
print(fruits)
```

*Adding tuple to a tuple*

```python
fruits = ("Apple","Banana","Cherry")
add_fruits = ("Orange",)
fruits += add_fruits
print(fruits)
```

*Removing items from a tuple*

```python
fruits = ("Apple", "Banana", "Mango", "cherry", "grapes")
print(type(fruits))
y = list(fruits)
print(type(y))

y.remove("cherry")
fruits = tuple(y)
print(fruits)
```

**Loop through tuple using the index numbers**

*   We can use a for loop with index numbers to access tuple items one by one.
*   len(fruits) returns the total number of items.
*   range() generates index numbers.
*   fruits[i] accesses each item using its index.

```python
fruits = ("Apple","Banana","Cherry")
for i in range(len(fruits)):
  print(fruits[i])
```

**Loop Through Tuple Using While Loop**

*   while loop mein range() ki zarurat nahi hoti. i khud index number ko control karta hai.
*   while loop → Counter variable (i) use karte hain.

```python
name = ("Apple", "Banana", "Cherry")
i = 0
while i < len(name):
  print(name[i])
  i = i+1
```
