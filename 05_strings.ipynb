**\n**: \n in Python is called a newline character. It is used to move the cursor to the next line.

```python
# \n
str1 = "My name is\nRoshan Yadav"
print(str1)
```

**\t**: \t in Python is called a Tab Character. It is used to give horizontal space (tab space) between text.

```python
str1 = "My name is\tRoshan Yadav"
print(str1)
```

**2. String Concatenation**: Concatenation means joining two or more strings together using the **+** operator.

```python
# simple
a = "roshan"
b = " is a good boy"
c = a + b
print(c)
```

```python
# string concatenation with space between them
str1 = "My name is"
str2 = "Roshan Yadav"
print(str1+" "+str2)
```

```python
# best way to combine two string.
str1 = "My name is"
str2 = "Roshan Yadav"

# Combine two strings and store them in a new variable.
strFinal = str1 + " " + str2

# Print the final string.
print(strFinal)
```

```python
# format unlimited arguments
qty = 3
itemno = 567
price = 49.95
myorder = "I want {} pieces of item number {} for price {} rupees."
print(myorder.format(qty, itemno, price))
```

```python
# format unlimited arguments with indexing.
qty = 3
itemno = 567
price = 49.95
myorder = "I want pay {2} rupees for {0} piceses of item no {1}."
print(myorder.format(qty, itemno, price))
```

**4. Indexing in Python**: Indexing is used to access individual characters of a string using their position number.

**5. Slicing in Python**: Slicing is used to extract a part of a string using index positions.

*   string[start index : end index : step]
*   The ending index is not counted.
*   Slicing does not change the original string. It only gives a new part of the string.

```python
str1 = "Roshan Yadav"
print(str1[1:4])
```

```python
print(str1[0:5])
```

```python
print(str1[:5])
```

```python
print(str1[1:])
```

```python
print(str1[:])
```

**Negative Indexing**: It is used to access a single character of a string from the end.

```python
name = "Roshan"
print(name[-1]) # n (Indexing)
```

```python
print(name[-2]) # a (Indexing)
```

**Negative Slicing**: It is used to get a part of a string using negative index numbers.

```python
name = "Roshan"
print(name[-3:-1])
```

```python
print(name[-1:-6:-1])
```

**6. String Functions/Methods**: These are built-in methods used to work with and modify strings.

```python
# capitalize()
str1 = "my name is roshan yadav"
print(str1.capitalize())
```

```python
# upper case
str1 = "my name is roshan yadav"
print(str1.upper())
```

```python
#lower case
str1 = "MY NAME IS ROSHAN YADAV"
print(str1.lower())
```

```python
#removing white spaces
x = "   Roshan, Yadav   "
print(x.strip())
```

```python
# Replace string
x = "He is Rohan"
print(x.replace("R", "S"))
print(x.replace("Rohan", "Roshan"))
```

```python
# split string
a = "Roshan Yadav"
print(a.split(",")) #output- array mein ayega
```

```python
# split string (best way)
a = "Roshan, Yadav"
b = a.split(",") #output- array mein ayega
print(b)
```

```python
#title()
txt = "I AM LEARNING PYTHON"
print(txt.title())
```

```python
# endwith
str1 = "my name is roshan yadav"
print(str1.endswith("yadav"))
```

```python
# endwith
str1 = "my name is roshan yadav"
print(str1.endswith("av"))
```

```python
# startwith()
name = "my name is roshan yadav"
print(name.startswith("my"))
print(name.startswith("roshan"))
```

**find()**: It returns the index of the first occurrence of a character or word in a string.

```python
# find()
name = "Python is easy and Python is powerful"
# Find the first occurrence of Python
print(name.find("Python"))
```

```python
# count()
name = "Python is easy and Python is powerful"

# is kitni baar aya hai name variable mein.
print(name.count("is"))
```
